## Project Overview

This project analyzes customer demographic and purchasing behavior data using Python, Pandas, NumPy, Matplotlib, and Seaborn.

## 1. Customer Age Distribution

```python
plt.figure(figsize=(8, 5))

plt.hist(df["age"], bins=10, edgecolor="black")

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()
```

## 2. Customer Distribution by Region

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="region")

plt.title("Customer Distribution by Region")
plt.xlabel("Region")
plt.ylabel("Number of Customers")

plt.show()
```

## 3. Annual Income Distribution

```python
plt.figure(figsize=(8, 5))

plt.hist(df["annual_income"], bins=10, edgecolor="black")

plt.title("Annual Income Distribution")
plt.xlabel("Annual Income")
plt.ylabel("Number of Customers")

plt.show()
```

## 4. Purchase Amount Distribution

```python
plt.figure(figsize=(8, 5))

plt.hist(df["purchase_amount"], bins=10, edgecolor="black")

plt.title("Purchase Amount Distribution")
plt.xlabel("Purchase Amount")
plt.ylabel("Number of Customers")

plt.show()
```

## 5. Loyalty Score Distribution

```python
plt.figure(figsize=(8, 5))

plt.hist(df["loyalty_score"], bins=8, edgecolor="black")

plt.title("Customer Loyalty Score Distribution")
plt.xlabel("Loyalty Score")
plt.ylabel("Number of Customers")

plt.show()
```

## 6. Purchase Frequency Distribution

```python
plt.figure(figsize=(8, 5))

plt.hist(df["purchase_frequency"], bins=10, edgecolor="black")

plt.title("Purchase Frequency Distribution")
plt.xlabel("Purchase Frequency")
plt.ylabel("Number of Customers")

plt.show()
```

## 7. Annual Income vs Purchase Amount

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="annual_income",
    y="purchase_amount"
)

plt.title("Annual Income vs Purchase Amount")
plt.xlabel("Annual Income")
plt.ylabel("Purchase Amount")

plt.show()
```

## 8. Loyalty Score vs Purchase Amount

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="loyalty_score",
    y="purchase_amount"
)

plt.title("Loyalty Score vs Purchase Amount")
plt.xlabel("Loyalty Score")
plt.ylabel("Purchase Amount")

plt.show()
```

## 9. Correlation Analysis

```python
correlation = df[
    ["age", "annual_income", "purchase_amount",
     "loyalty_score", "purchase_frequency"]
].corr()

correlation
```

## 10. Correlation Heatmap

```python
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Customer Variables")
plt.show()
```

## 11. Key Findings

### Key Findings

- The dataset contains 238 customer records across 7 variables.
- Customer ages range from 22 to 55 years, with an average age of approximately 39 years.
- Annual income ranges from 30,000 to 75,000, with an average of approximately 57,408.
- Purchase amounts range from 150 to 640, with an average of approximately 426.
- The average customer loyalty score is approximately 6.79.
- Purchase frequency ranges from 10 to 28 purchases, with an average of approximately 19.8.
- The correlation analysis shows very strong positive relationships among age, annual income, purchase amount, loyalty score, and purchase frequency.
- Purchase amount has a very strong positive relationship with both loyalty score and purchase frequency.
- Regional distribution helps identify how customers are spread across different regions.

## 12. Business Insights

- Customers with higher purchase amounts tend to have higher loyalty scores.
- Customers with higher purchase amounts also tend to have higher purchase frequencies.
- Income and age show a strong positive relationship with purchasing behavior in this dataset.
- Businesses can use customer purchasing patterns to identify valuable customer segments.
- Loyalty and purchase frequency can be useful indicators when analyzing customer engagement.
- Regional customer distribution can help businesses understand where their customers are concentrated.
